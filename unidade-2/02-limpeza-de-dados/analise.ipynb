# Limpeza de dados de automóveis

Padronizar valores ausentes, tipos numéricos e categorias de uma base de automóveis com uma classe reutilizável.

Execute as células em ordem. Este notebook foi revisado a partir de uma atividade acadêmica para apresentação em portfólio.

## Exploração inicial

```python
import pandas as pd #import do pandas

df = pd.read_csv('dados/Automobile_data.csv') #import do repositório csv

display(df.head()) #exibe as cinco primeiras linhas

print("-" * 60)

df.info() #exibe as informações estruturais

print("-" * 60)

print(df.shape)  # exibe as dimensões do dataset, linhas x coluna

print("-" * 60)

print(df.dtypes) # exibe os tipos de dados
```

## Classe DataCleaning

```python
class DataCleaning:

    def __init__(self, df):
        self.df = df.copy() #construtor

    def null_standardization(self):
        self.df = self.df.replace(['?', '', 'N/A', '-', 'null'], pd.NA)  #padronização de valores nulos

    def convert_number(self):
        colunas = ['normalized-losses', 'bore', 'stroke',  'horsepower', 'peak-rpm', 'price']

        for coluna in colunas:
            self.df[coluna] = pd.to_numeric(self.df[coluna], errors='coerce')  #conversão colunas strings para números

    def null_analysis(self):
        nulos = self.df.isna().sum()

        resultado = pd.DataFrame({'Nulos': nulos, 'Percentual (%)': (nulos / len(self.df) * 100).round(2)})

        print(resultado)
        print('Linhas incompletas:', self.df.isna().any(axis=1).sum())
        print('Linhas inteiramente nulas:', self.df.isna().all(axis=1).sum()) #análise da quantidade e do percentual de nulos

    def text_pattern(self, case='lower'):
        colunas = self.df.select_dtypes(include=['object', 'string']).columns

        for coluna in colunas:
            if case == 'lower':
                self.df[coluna] = self.df[coluna].str.lower()
            elif case == 'upper':
                self.df[coluna] = self.df[coluna].str.upper()   #converte strings para minusculas ou maiusculas

    def category_validation(self):
        self.df['make'] = self.df['make'].replace({'alfa-romero': 'alfa-romeo', 'peugot': 'peugeot'}) # correção de categorias com nomes incorretos
```

## Aplicação e diagnóstico da limpeza

```python
limpeza = DataCleaning(df) # instanciação da classe

limpeza.null_standardization() # chamada de método
limpeza.convert_number() # chamada de método
limpeza.text_pattern(case='lower') # chamada de método
limpeza.category_validation() # chamada de método
limpeza.null_analysis() # chamada de método

df = limpeza.df # atualiza o dataframe com os dados tratados

display(df.head()) # exibe as cinco primeiras linhas
```

## Limites da limpeza

Valores ausentes são diagnosticados e preservados: não foi aplicada imputação nem exclusão de registros. As correções de categorias são específicas desta base. A coerção numérica converte valores inválidos em ausentes; o diagnóstico final permite acompanhar esse efeito.
