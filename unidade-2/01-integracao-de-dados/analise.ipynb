# Integração de multas ambientais

Comparar a arrecadação de multas ambientais de RN e PB por tipo de infração, integrando duas bases do IBAMA.

Execute as células em ordem. Este notebook foi revisado a partir de uma atividade acadêmica para apresentação em portfólio.

## Carregamento das bases

```python
# Leitura dos arquivos para DataFrames

import pandas as pd

df_rn = pd.read_csv('dados/arrecadacao_de_multas_ambientais_por_bens_tutelados_RN.csv', sep=";") # leitura do csv do rn
df_pb = pd.read_csv('dados/arrecadacao_de_multas_ambientais_por_bens_tutelados_PB.csv', sep=";") # leitura do csv da pb
```

## Exploração, conversão monetária e agregação

```python
display(df_rn[["UF", "Tipo Infração", "Valor Pago"]].head()) # exibição das 5 primeiras linhas do dataframe do RN
df_rn.info() # exibição de info do dataframes

print("-" * 60)

display(df_pb[["UF", "Tipo Infração", "Valor Pago"]].head()) # exibição das 5 primeiras linhas do dataframe da PB
df_pb.info() # exibição de info do dataframes

df_rn["Valor Pago"] = (df_rn["Valor Pago"].str.replace(".", "", regex=False).str.replace(",", ".", regex=False).astype(float)) # conversão de valor pago de string para float, assim como tratamento para formatação de moeda BRL

df_pb["Valor Pago"] = (df_pb["Valor Pago"].str.replace(".", "", regex=False).str.replace(",", ".", regex=False).astype(float))  # conversão de valor pago de string para float, assim como tratamento para formatação de moeda BRL

arrecadacao_rn = (df_rn.groupby("Tipo Infração", as_index=False)["Valor Pago"].sum()) # agregação por tipo de infração, assim como a soma dos valores no dataframe do RN

arrecadacao_pb = (df_pb.groupby("Tipo Infração", as_index=False)["Valor Pago"].sum()) # agregação por tipo de infração, assim como a soma dos valores no dataframe da PB

display(arrecadacao_rn)
display(arrecadacao_pb)
```

## Integração por tipo de infração

```python
df_integrado = pd.merge(arrecadacao_rn, arrecadacao_pb, on="Tipo Infração", how="outer", suffixes=("_RN", "_PB"), validate="one_to_one")  # integra RN e PB pela coluna "Tipo Infração"

display(df_integrado) # exibe o DataFrame integrado

df_integrado.size # retorna o número total de elementos do DataFrame
```

## Análise de completude

```python
print(df_integrado.isna().sum()) # conta os valores ausentes em cada coluna

print("Total de dados ausentes:", df_integrado.isna().sum().sum()) # conta o total de valores ausentes no DataFrame
```

## Interpretação

Cada registro representa um pagamento/parcela: a soma considera os valores pagos, não o valor nominal das multas. O `outer merge` conserva os tipos de infração encontrados em qualquer estado. Valores ausentes após a integração não são automaticamente substituídos por zero. A análise é descritiva e não permite comparar desempenho de fiscalização sem considerar população, período e contexto.
