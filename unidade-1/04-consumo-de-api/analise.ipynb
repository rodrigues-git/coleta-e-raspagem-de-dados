# Consumo da API Rick and Morty

Consultar personagens e transformar respostas JSON em tabelas, incluindo uma consulta à página 2 com status alive.

Execute as células em ordem. Este notebook foi revisado a partir de uma atividade acadêmica para apresentação em portfólio.

## Consulta de personagens

```python
import requests
import pandas as pd

url = "https://rickandmortyapi.com/api/character" # declaração de endpoint

resposta = requests.get(url, timeout=30) # requisição para api

resposta.raise_for_status()
resposta_json = resposta.json() # conversão para json

characters = resposta_json.get("results", []) # selecionando apenas as informações de "results" para dataframe

df = pd.json_normalize(characters) # conversão para dataframe

#print de resposta
print("5 primeiros registros")
df.head()
```

## Filtro por página e status

```python
parametros = {
    "page": 2,
    "status": "alive"
} # definição dos parametros para a  requisição

resposta = requests.get(url, params=parametros, timeout=30) # requisição com os parametros

if resposta.status_code == 404:
    resposta_json = {"results": []}
else:
    resposta.raise_for_status()
    resposta_json = resposta.json() # conversão da resposta da requisição para json

characters = resposta_json.get("results", [])  # à principio, seleciona apenas as informações de "results", caso não exista, isto é, não existem registros para os filtros especificados, retorna uma lista vazia

if len(characters) > 0: # verifica tamanho de characters para validação se encontrou personagens

  df  = pd.json_normalize(characters) # criação de dataframe

  # exibição de resultado
  print("5 primeiros registros")
  display(df.head())

else:
  print("Não foram encontrados personagens para os filtros especificados")
```
