# Leitura de TXT, CSV e PDF

Ler fontes em formatos diferentes e produzir indicadores de avaliações e vendas.

## Restaurantes em TXT

```python
with open("dados/restaurantes.txt", "r", encoding="utf-8") as file:
  content = file.readlines() #abre o arquivo e armazena cada linha no array "content"

restaurantes = [] # declaração de array vazio onde ficarão os restaurantes

for linha in content:
  nome, cidade, avaliacao = linha.strip().split(";") # para cada linha em content, armazena de acordo com o delimitador ";", remove espaços em branco e elimina quebras de linha (\n) no final do texto

  restaurante = {
      "nome": nome,
      "cidade": cidade,
      "avaliacao": float(avaliacao)
  } # criação de estrutua para armazanemanto de cada restaurante

  restaurantes.append(restaurante) #adiciona o restaurante da vez na lista

restaurantes_natal = [] # declaração de array vazio onde ficarão os restaurantes de natal

#para cada restaurante do array restaurantes, verifica se a cidade é natal, se for, adiciona no array de restaurante de natal
for restaurante in restaurantes:
  if restaurante["cidade"] == "Natal":
    restaurantes_natal.append(restaurante)
```

```python
soma = 0 # declaração de variável que armazenará a soma das avaliações

# para cada restaurante no array restaurantes_natal, adiciona sua avaliação ao somador e imprime seu nome e avaliação individual
for restaurante in restaurantes_natal:

  print(f"{restaurante['nome']} - {restaurante['avaliacao']}")

  soma = soma + restaurante['avaliacao']

media = soma / len(restaurantes_natal) if restaurantes_natal else float("nan") #calculo da média

print(f"Avaliação Média: {media}") #print da média
```

## Leitura com a biblioteca csv

```python
import csv

restaurantes_natal = [] # declaração de array vazio onde ficarão os restaurantes de natal

with open("dados/restaurantes.csv", "r", encoding="utf-8") as file:
    content = csv.DictReader(file) # lê o arquivo e armazena como dicionario em content

    # percorre content e verifica se a cidade é Natal, se for, converte o campo avaliação para float e adiciona no array de restaurantes de Natal
    for restaurante in content:
      if(restaurante["cidade"] == "Natal"):
        restaurante["avaliacao"] = float(restaurante["avaliacao"])
        restaurantes_natal.append(restaurante)

soma  = 0 # declaração de variável que armazenará a soma das avaliações

# para cada restaurante no array restaurantes_natal, adiciona sua avaliação ao somador
for restaurante in restaurantes_natal:
  soma = soma + restaurante["avaliacao"]

#calculo da média
media = soma / len(restaurantes_natal) if restaurantes_natal else float("nan")

#print da média
print(f"Avaliação Média: {media}")
```

## Leitura com pandas

```python
import pandas as pd

df = pd.read_csv("dados/restaurantes.csv")  # lê o arquivo e já cria um dataframe df

restaurantes_natal = df[df["cidade"] == "Natal"] # cria novo dataframe apenas com os restaurantes de Natal

media = restaurantes_natal["avaliacao"].mean() # calculo da media com função mean

print(f"Média das avaliações: {media}") # print da media
```

Considero a abordagem com pandas mais conveniente, pois ela é mais direta, simples e fácil de utilizar. A biblioteca possui funções próprias para operações comuns, como o cálculo da média, o que facilita bastante a análise dos dados. Além disso, é possível criar um DataFrame diretamente a partir da leitura do arquivo, tornando o processo mais prático e organizado. Dessa forma, todos os dados do arquivo ficam armazenados no DataFrame, permitindo realizar  filtros, cálculos e outras análises sem perder as informações originais.

## Extração de vendas de PDF

```python
from pypdf import PdfReader

path = "dados/relatorio_vendas.pdf" #caminho do arquivo pdf

reader = PdfReader(path) #lê o arquivo pdf

texto = "\n".join(page.extract_text() or "" for page in reader.pages) # Lê todas as páginas

texto = " ".join(texto.split()) #remove espaços e quebras de linha extras do texto extraído

produtos = [] #declaração de array vazio onde ficarão armazenados os produtos

partes = texto.split("Produto:") #separa o texto sempre que encontrar "Produto:"

for parte in partes[1:]: #percorre cada produto encontrado no texto

    nome, restante = parte.split("Quantidade:") #separa o nome do produto do restante das informações

    quantidade, valor_unitario = restante.split("Valor unitário:") # separa a quantidade do valor unitário

    # remove espaços extras dos valores
    nome = nome.strip()
    quantidade = quantidade.strip()
    valor_unitario = valor_unitario.strip()

    # converte a quantidade para inteiro e o valor unitário para float
    quantidade = int(quantidade)
    valor_unitario = float(valor_unitario)

    valor_total = quantidade * valor_unitario #calcula o valor total do produto

    # cria um dicionário com as informações do produto
    produto = {
        "produto": nome,
        "quantidade": quantidade,
        "valor_unitario": valor_unitario,
        "valor_total": valor_total
    }

    produtos.append(produto) #adiciona o produto no array de produtos

total_vendas = 0 # variável que armazenará o valor total de todas as vendas

maior_venda = produtos[0] # inicialmente considera o primeiro produto como a maior venda

# percorre cada produto armazenado
for produto in produtos:

    # adiciona o valor total do produto ao total das vendas
    total_vendas = total_vendas + produto["valor_total"]

    # verifica se o produto atual possui valor total maior que a maior venda
    if produto["valor_total"] > maior_venda["valor_total"]:
        maior_venda = produto

# imprime as informações de cada produto
for produto in produtos:

    print(produto["produto"])
    print(f"Quantidade: {produto['quantidade']}")
    print(f"Valor unitário: R$ {produto['valor_unitario']:.2f}".replace(".", ","))
    print(f"Total: R$ {produto['valor_total']:.2f}".replace(".", ","))
    print("-" * 40)

print(f"Total de vendas: R$ {total_vendas:.2f}".replace(".", ",")) #print do valor total das vendas

# print do produto com maior valor total
print(f"Maior venda: {maior_venda['produto']}") #print do valor total das vendas
```
