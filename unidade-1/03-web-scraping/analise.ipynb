# Web scraping de citações

Extrair a primeira citação, seu autor e a tag HTML do texto no site de demonstração Quotes to Scrape.

Execute as células em ordem. Este notebook foi revisado a partir de uma atividade acadêmica para apresentação em portfólio.

## Coleta e interpretação de HTML

```python
path = "https://quotes.toscrape.com/"

import requests # import da dependência
from bs4 import BeautifulSoup #import da classe beautifulsoup de bs4

response = requests.get(path, timeout=30) # realiza requisição na url especificada em path

print(response.status_code) # imprime status code da requisição, a fim de verificar seu êxito

response.raise_for_status()

content = response.content # extrai o conteudo da requisição

site = BeautifulSoup(content, 'html.parser') # transforma o conteudo da requisição em uma estrutura manipulavel e navegavel pelo python.

citacao = site.find('div', attrs={'class': 'quote'}) #extrai a primeira div com a classe quote

if citacao is None:
    raise ValueError('Nenhuma citação encontrada; verifique a estrutura da página.')

citacao_texto = citacao.find('span', attrs={'class': 'text'}) #dentro da div com classe quote, extrai a primeira tag span com class text

citacao_autor = citacao.find('small', attrs={'class': 'author'}) #dentro da div com classe quote, extrai a primeira tag com class author

if citacao_texto is None or citacao_autor is None:
    raise ValueError("A citação não contém os elementos de texto e autor esperados.")
```

```python
print("Citação:", citacao_texto.text) # imprime a citação

print("Autor:", citacao_autor.text) # imprime o autor

print("Tag do texto:", citacao_texto.name) # imprime a tag da citacao
```
